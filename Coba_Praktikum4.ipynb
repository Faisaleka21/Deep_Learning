```python
# IMPORT PYTHON
import torch
import torch.nn as nn
```

```python
# 3 input -> 1 output
model = nn.Linear(3,1)
print (model)
```

```python
# ambil 1 data contoh (rumah 1)
x = torch.tensor([[60.0, 2.0, 25.0], #[1, 1]
                  [80.0, 3.0, 10.0], #[2, 2]
                  [120.0, 4.0, 5.0]  #[3, 3]
                  ])

# forward pss
z = model (x)           # output liner
pred = torch.relu(z)    # aktivasi relu

print ("Nili z (sebelum aktivasi): ",z)
print ("Nilai pred (sesudah ReLu): ",pred)
```

```python
# target harg 3 rumah (juta rupiah)
target = torch.tensor ([
    [600.0],
    [800.0],
    [1200.0]
    ])# bentuk[1, 1]

# hitung loss (MSE)
criterion = nn.MSELoss()
loss = criterion(pred, target)
print ("Prediksi: ", pred) # jika ditambakan .item itu untuk satu elemen/grup saja
print ("Target: ", target)
print ("Nilai Loss (MSE): ",loss.item())
```

```python
# reset gradiant lma
model.zero_grad()

# backward pass / analisa kembali
loss.backward()

# Lihat grdiant
print("Gradiant Weight (dl/dw) : ")
print(model.weight.grad)
print("Gradiant Bias (dl/db) : ")
print(model.bias.grad)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
print("Weight sebelum update: ", model.weight.data)
print("Bias sebelum update: ", model.bias.data)
optimizer.step()
print("Weight setelah update: ", model.weight.data)
print("Bias setelah update: ", model.bias.data)
```
